Eine Schwingung ist eine **zeitlich periodische Änderung** eines physikalischen Zustands.
Für die Auslenkung gilt:
$$A(t + T) = A(t)$$

Dabei ist:
- $T$ die Periodendauer,
- $A(t)$ die momentane Auslenkung aus der Ruhelage.

- $\varphi$ : Phasenkonstante  

Die Phase $(\omega t + \varphi)$ beschreibt den momentanen Schwingungszustand.

## Grundlagen für das 2. Experiment: Erzwungene Schwingungen und Resonanz (Drehpendel)

### 1. Erzwungene Schwingung
Wirkt auf ein schwingungsfähiges System zusätzlich zu Rückstellkraft und Dämpfung eine
**zeitlich periodische äußere Kraft**, so spricht man von einer erzwungenen Schwingung.

Die äußere Kraft habe die Form:
$$F(t) = F_0 \cdot cos(\omega t)$$

Dabei ist:
- $F_0$ die Amplitude der äußeren Kraft,
- $\omega$ die Kreisfrequenz der Anregung.

Nach einer Einschwingzeit führt das System eine stationäre Schwingung aus, deren Frequenz
mit der Anregungsfrequenz identisch ist.

---

### 2. Bewegungsgleichung der erzwungenen Schwingung
Die Bewegungsgleichung des gedämpften, erzwungenen Systems lautet:

$$m \cdot \ddot{x} + k \cdot \dot{x} + D \cdot x = F_0 \cdot cos(\omega t)$$

Dabei stehen die Terme für:
- Trägheitskraft ($m \cdot \ddot{x}$),
- Reibungskraft ($k \cdot \dot{x}$),
- Rückstellkraft ($D \cdot x$),
- äußere periodische Kraft ($F_0 \cdot cos(\omega t)$).

---

### 3. Stationäre Lösung
Nach dem Abklingen des Einschwingvorgangs ergibt sich eine stationäre Lösung der Form:

$$x(t) = x_0(\omega) \cdot cos(\omega t + \varphi)$$

Dabei sind:
- $x_0(\omega)$ die frequenzabhängige Amplitude,
- $\varphi$ die Phasenverschiebung zwischen Anregung und Schwingung.

---

### 4. Frequenzabhängigkeit der Amplitude (Resonanzkurve)
Die Amplitude der erzwungenen Schwingung hängt von der Anregungsfrequenz ab:

$$x_0^2(\omega) =
\frac{F_0^2}{m^2\left[(\omega_0^2 - \omega^2)^2 + (2\delta\omega)^2\right]}$$

mit:
- $\omega_0$ der Eigenkreisfrequenz,
- $\delta$ der Dämpfungskonstante.

Trägt man $x_0$ gegen $\omega$ auf, erhält man die **Resonanzkurve**.

---

### 5. Resonanz
Die Resonanz ist der Zustand, bei dem die Amplitude der erzwungenen Schwingung maximal ist.
Die Resonanzfrequenz liegt in der Nähe der Eigenfrequenz des Systems, ist jedoch bei
vorhandener Dämpfung leicht verschoben.

Im Resonanzfall:
- ist die Energieaufnahme des Systems maximal,
- erreicht die Amplitude ihr größtes Ausmaß.

---

### 6. Phasenverschiebung
Zwischen äußerer Kraft und Antwort des Systems besteht eine frequenzabhängige
Phasenverschiebung:

$$tan(\varphi) =
\frac{2\delta\omega}{\omega^2 - \omega_0^2}$$

Charakteristische Fälle:
- kleine Frequenzen: $\varphi \approx 0$ (in Phase),
- Resonanznähe: $\varphi \approx \frac{\pi}{2}$,
- große Frequenzen: $\varphi \approx \pi$ (gegenphasig).

---

### 7. Halbwertsbreite
Zur quantitativen Beschreibung der Resonanz wird die **Halbwertsbreite** verwendet.

Definition:
- Man bestimmt die Frequenzen, bei denen gilt:
$$x = \frac{x_{max}}{\sqrt{2}}$$
- Der Frequenzabstand dieser beiden Punkte ist die Halbwertsbreite $\Delta\omega$.

Zwischen Halbwertsbreite und Dämpfung gilt näherungsweise:
$$\Delta\omega = 2\delta$$

---

### 8. Gütefaktor
Der **Gütefaktor** beschreibt die Schärfe der Resonanz und die Energieverluste des Systems:

$$Q \approx \frac{\omega_0}{2\delta} = \frac{\omega_0}{\Delta\omega}$$

Ein großer Gütefaktor bedeutet:
- geringe Dämpfung,
- scharfe Resonanz,
- langsamen Energieverlust.

---

### 9. Bedeutung für das Experiment
Im 2. Experiment werden:
- die Resonanzkurve $x_0(\omega)$ aufgenommen,
- die Dämpfungskonstante $\delta$ aus der Halbwertsbreite bestimmt,
- der Gütefaktor $Q$ berechnet,
- die Ergebnisse mit jenen aus der freien Schwingung verglichen.

Damit wird der Zusammenhang zwischen freier und erzwungener Schwingung experimentell
bestätigt.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# ============================================================
# DATEN HIER EINTRAGEN (Resonanzmessung)
# ============================================================

# Anregungsfrequenzen f [Hz]
f = np.array([228, 260, 300, 305, 315, 320, 325, 330, 345, 370, 383]) / 600   # <-- ersetzen

# Gemessene Amplituden x (Skalenteile)
x = np.array([0.8, 1.1, 2.3, 2.6, 4.0, 4.8, 4.9, 4.7, 2.6, 1.5, 1.3])        # <-- ersetzen

# ============================================================
# HILFSFUNKTIONEN
# ============================================================

def format_wert(wert, uns, einheit):
    if not np.isfinite(uns) or uns == 0:
        return f"({wert:.6g}) {einheit}"
    exp = int(np.floor(np.log10(abs(uns))))
    decimals = max(0, -exp)
    wert_r = round(wert, decimals)
    uns_r = round(uns, decimals)
    fmt = f"{{:.{decimals}f}}"
    return f"({fmt.format(wert_r)} +- {fmt.format(uns_r)}) {einheit}"

# ============================================================
# 1) Resonanzfrequenz bestimmen (aus Messmaximum)
# ============================================================

idx_max = np.argmax(x)
f_res = f[idx_max]
x_max = x[idx_max]

# Unsicherheit: halber Frequenzabstand der Nachbarpunkte
if 0 < idx_max < len(f) - 1:
    f_res_uns = 0.5 * (f[idx_max + 1] - f[idx_max - 1])
else:
    f_res_uns = np.nan

w_res = 2 * np.pi * f_res
w_res_uns = 2 * np.pi * f_res_uns

# ============================================================
# 2) Halbwertsbreite bestimmen (Interpoliert)
# ============================================================

half_value = x_max / np.sqrt(2)

f1 = np.nan
f2 = np.nan

# linke Seite
for i in range(idx_max - 1, -1, -1):
    if (x[i] - half_value) * (x[i + 1] - half_value) < 0:
        f1 = f[i] + (half_value - x[i]) * (f[i+1] - f[i]) / (x[i+1] - x[i])
        break

# rechte Seite
for i in range(idx_max, len(f) - 1):
    if (x[i] - half_value) * (x[i + 1] - half_value) < 0:
        f2 = f[i] + (half_value - x[i]) * (f[i+1] - f[i]) / (x[i+1] - x[i])
        break

if not np.isfinite(f1) or not np.isfinite(f2):
    raise ValueError("Halbwertsstellen konnten nicht bestimmt werden. Prüfe Datenbereich um das Maximum.")

delta_f = f2 - f1
delta_w = 2 * np.pi * delta_f

# Unsicherheit: halber Abstand der benachbarten Messpunkte (grobe Abschätzung)
delta_f_uns = 0.5 * np.mean(np.diff(f))
delta_w_uns = 2 * np.pi * delta_f_uns

# ============================================================
# 3) Dämpfungskonstante (Halbwertsmethode)
# ============================================================

damping_const = delta_w / 2
damping_uns = delta_w_uns / 2

# ============================================================
# 4) Gütefaktor (Halbwertsmethode)
# ============================================================

Q = w_res / (2 * damping_const)
Q_uns = Q * np.sqrt(
    (w_res_uns / w_res)**2 +
    (damping_uns / damping_const)**2
)

# ============================================================
# 5) Kurvenfit der Resonanzkurve (physikalisches Modell)
# ============================================================

def resonance_model(f, A0, f0, beta):
    """
    Modell für die Amplitude einer erzwungenen, gedämpften Schwingung:
    A(f) = A0 / sqrt((w0^2 - w^2)^2 + (2*beta*w)^2)
    f in Hz, beta in 1/s
    """
    w = 2 * np.pi * f
    w0 = 2 * np.pi * f0
    return A0 / np.sqrt((w0**2 - w**2)**2 + (2 * beta * w)**2)

# Startwerte (wichtig für stabilen Fit)
A0_guess = x_max * (2 * np.pi * f_res)**2
f0_guess = f_res
beta_guess = damping_const

p0 = [A0_guess, f0_guess, beta_guess]

# Optional: Grenzen (verhindert unphysikalische Werte)
bounds = ([0, 0, 0], [np.inf, np.inf, np.inf])

params, cov = curve_fit(resonance_model, f, x, p0=p0, bounds=bounds)

A0_fit, f0_fit, beta_fit = params
A0_uns, f0_uns, beta_uns = np.sqrt(np.diag(cov))

w0_fit = 2 * np.pi * f0_fit
w0_uns = 2 * np.pi * f0_uns

Q_fit = w0_fit / (2 * beta_fit)
Q_fit_uns = Q_fit * np.sqrt(
    (w0_uns / w0_fit)**2 +
    (beta_uns / beta_fit)**2
)

# ============================================================
# AUSGABE (nummeriert, korrekt formatiert)
# ============================================================

print("Auswertung Experiment 2 (PW12) – erzwungene Schwingung\n")

print("1) Resonanzfrequenz (aus Maximum)")
print("   w_res =", format_wert(w_res, w_res_uns, "rad/s"))

print("\n2) Halbwertsbreite")
print("   Delta_w =", format_wert(delta_w, delta_w_uns, "rad/s"))

print("\n3) Daempfungskonstante (Halbwertsmethode)")
print("   beta =", format_wert(damping_const, damping_uns, "1/s"))

print("\n4) Guetefaktor (Halbwertsmethode)")
print("   Q =", format_wert(Q, Q_uns, ""))

print("\n5) Kurvenfit (Resonanzfunktion)")
print("   w0 =", format_wert(w0_fit, w0_uns, "rad/s"))
print("   beta_fit =", format_wert(beta_fit, beta_uns, "1/s"))
print("   Q_fit =", format_wert(Q_fit, Q_fit_uns, ""))

# ============================================================
# PLOT (Messwerte + Fit + Halbwertslinien)
# ============================================================

f_fine = np.linspace(min(f), max(f), 500)
x_fit = resonance_model(f_fine, *params)

plt.figure(figsize=(7.5, 4.6))
plt.errorbar(f*2*np.pi, x, yerr=0.2, xerr=0.157, fmt="o", label="Messwerte")
plt.plot(f_fine*2*np.pi, x_fit, "-", label="Fit")
plt.axhline(half_value, linestyle="--", label="x_max / sqrt(2)")
plt.axvline(f1*2*np.pi, linestyle=":", color="gray")
plt.axvline(f2*2*np.pi, linestyle=":", color="gray")
plt.xlabel("Kreisfrequenz omega [rad/s]")
plt.ylabel("Amplitude x")
plt.title("Resonanzkurve der erzwungenen Schwingung")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

## Grundlagen – Gekoppelte Schwingungen (Schwebungsfall)

### 1. Physikalisches System
Zwei **identische Pendel** sind über eine schwache Feder gekoppelt.  
Wird **nur eines** der Pendel ausgelenkt, so tauschen die Pendel periodisch Energie aus.
Dieser periodische Energieaustausch heißt **Schwebung**.

Man beobachtet:
- eine **schnelle Schwingung** (ähnlich wie beim einzelnen Pendel)
- eine **langsame Hüllkurve** (An- und Abschwellen der Amplitude)

---

### 2. Eigenmoden des Systems
Das gekoppelte System besitzt **zwei Eigenfrequenzen**:

- $\omega_0$: gleichphasige Schwingung  
  → Kopplungsfeder wird nicht gedehnt  
- $\omega_1$: gegenphasige Schwingung  
  → Kopplungsfeder wird maximal gedehnt  

Diese Eigenfrequenzen bestimmen vollständig das Bewegungsverhalten.

---

### 3. Schwebung als Überlagerung
Die Bewegung eines Pendels im Schwebungsfall ist die Überlagerung beider Eigenmoden:

$$
x(t) = A \cos(\omega_2 t)\,\cos(\omega_S t)
$$

Dabei gilt:

- $\omega_2$: schnelle Kreisfrequenz („Trägerfrequenz“)
- $\omega_S$: Schwebungs-Kreisfrequenz (Hüllkurve)

mit den Definitionen:

$$
\omega_2 = \frac{\omega_1 + \omega_0}{2}
\qquad
\omega_S = \frac{\omega_1 - \omega_0}{2}
$$

---

### 4. Schwebungsdauer
Die **Schwebungsdauer** ist die Zeit zwischen zwei Stillständen desselben Pendels:

$$
T_S = \frac{\pi}{\omega_S}
\qquad\Rightarrow\qquad
\omega_S = \frac{\pi}{T_S}
$$


---

### 5. Schnelle Schwingungsdauer
Die schnelle Schwingung besitzt die Periodendauer:

$$
T_2 = \frac{2\pi}{\omega_2}
\qquad\Rightarrow\qquad
\omega_2 = \frac{2\pi}{T_2}
$$

Sie wird gemessen, indem man **viele Perioden** zählt und mittelt.

---

### 6. Bestimmung der Eigenfrequenzen
Aus den gemessenen Größen $\omega_S$ und $\omega_2$ erhält man:

$$
\omega_1 = \omega_2 + \omega_S
\qquad
\omega_0 = \omega_2 - \omega_S
$$

---

### 7. Kopplungsgrad
Der **Kopplungsgrad** beschreibt die Stärke der Kopplung zwischen den Pendeln:

$$
K = \frac{D^*}{D + D^*}
$$

Experimentell lässt sich $K$ direkt aus den Frequenzen bestimmen:

$$
K = \frac{\omega_1^2 - \omega_0^2}{\omega_1^2 + \omega_0^2}
= \frac{2\,\omega_S\,\omega_2}{\omega_S^2 + \omega_2^2}
$$

---

### 8. Physikalische Interpretation
- $K \to 0$: sehr schwache Kopplung → kaum Energieübertragung  
- größeres $K$: schnellerer Energieaustausch → kürzere Schwebungsdauer  

Die Schwebung ist ein direktes Experiment zur **Beobachtung von Eigenmoden**.

---

### 9. Ziel der Auswertung
In der Auswertung werden bestimmt:

1. Schwebungsdauer $T_S$
2. Schnelle Periodendauer $T_2$
3. Eigenfrequenzen $ \omega_0, \omega_1 $
4. Kopplungsgrad $K$

Alle Größen werden inklusive **Messunsicherheiten** angegeben.


```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# DATEN HIER EINTRAGEN (Schwebungsfall)
# ============================================================

# Gemessene Zeiten fuer N schnelle Schwingungen (omega2)
t_start_fast = 0.47      # s
t_end_fast   = 52.18   # s
N_fast       = 29      # Anzahl Perioden

# Gemessene Zeiten fuer M Schwebungen (Stillstand -> Stillstand)
t_start_slow = 0.47      # s
t_end_slow   = 52.18   # s
M_slow       = 1        # Anzahl Schwebungen

# ============================================================
# HILFSFUNKTIONEN
# ============================================================

def freq_aus_zeitspanne(t_start, t_end, N):
    """
    Berechnet Kreisfrequenz w = 2*pi/T
    Unsicherheit aus einer Periodenmessung
    """
    T = (t_end - t_start) / N
    T_uns = T / np.sqrt(N)       # einfache statistische Schaetzung
    w = np.pi / T
    w_uns = w * (T_uns / T)
    return w, w_uns, T, T_uns

def format_wert(wert, uns, einheit):
    if not np.isfinite(uns) or uns == 0:
        return f"({wert:.6g}) {einheit}"
    exp = int(np.floor(np.log10(abs(uns))))
    decimals = max(0, -exp)
    wert_r = round(wert, 2)
    uns_r = round(uns, 2)
    fmt = f"{{:.{2}f}}"
    return f"({fmt.format(wert_r)} +- {fmt.format(uns_r)}) {einheit}"

# ============================================================
# 1) Frequenzen aus Messungen
# ============================================================

# schnelle Schwingung: omega_2
w2, w2_uns, T2, T2_uns = freq_aus_zeitspanne(
    t_start_fast, t_end_fast, N_fast
)

# Schwebung: omega_S
wS, wS_uns, TS, TS_uns = freq_aus_zeitspanne(
    t_start_slow, t_end_slow, M_slow
)

# ============================================================
# 2) Eigenfrequenzen omega0 und omega1
# ============================================================

w1 = w2 + wS
w0 = w2 - wS

w1_uns = np.sqrt(w2_uns**2 + wS_uns**2)
w0_uns = np.sqrt(w2_uns**2 + wS_uns**2)

# ============================================================
# 3) Kopplungsgrad K (Gleichung 21)
# ============================================================

K = (w1**2 - w0**2) / (w1**2 + w0**2)

# Fehlerfortpflanzung (linearisiert)
K_uns = K * np.sqrt(
    (2 * w1 * w1_uns / (w1**2 - w0**2))**2 +
    (2 * w0 * w0_uns / (w1**2 - w0**2))**2 +
    (2 * w1 * w1_uns / (w1**2 + w0**2))**2 +
    (2 * w0 * w0_uns / (w1**2 + w0**2))**2
)

# ============================================================
# AUSGABE
# ============================================================

print("Auswertung Experiment 3 (PW12) - gekoppelten Schwingungen\n")

print("1) Charakteristische Kreisfrequenzen")
print("   w2 (schnelle Schwingung) =", format_wert(w2, w2_uns, "rad/s"))
print("   wS (Schwebung)           =", format_wert(wS, wS_uns, "rad/s"))

print("\n2) Eigenfrequenzen des Systems")
print("   w0 (gleichsinnig) =", format_wert(w0, w0_uns, "rad/s"))
print("   w1 (gegensinnig)  =", format_wert(w1, w1_uns, "rad/s"))

print("\n3) Kopplungsgrad")
print("   K =", format_wert(K, K_uns, ""))


```
