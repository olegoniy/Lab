# Grundlagen zu Wechselstrom und Oszilloskopmessungen

In diesem Versuch werden ausschließlich sinusförmige Wechselspannungen und Wechselströme betrachtet. Spannung und Strom folgen dabei einer Sinusfunktion:

$$
U(t) = U_0 \sin(\omega t + \varphi_U), \qquad
I(t) = I_0 \sin(\omega t + \varphi_I).
$$

Die Kreisfrequenz hängt mit der Frequenz zusammen:

$$
\omega = 2\pi\nu,
$$

und die Periodendauer ist

$$
T = \frac{1}{\nu}.
$$

Zwischen Spannung und Strom gibt es eine Phasenverschiebung:

$$
\Delta\varphi = \varphi_U - \varphi_I.
$$

Ist $\Delta\varphi < 0$, eilt der Strom voraus; ist $\Delta\varphi > 0$, hinkt der Strom der Spannung hinterher.

---

## Effektivwerte

Digitale Multimeter messen Effektivwerte (RMS). Für Sinussignale gilt:

$$
U_\text{eff} = \frac{U_0}{\sqrt{2}}, \qquad
I_\text{eff} = \frac{I_0}{\sqrt{2}}.
$$

Ein Oszilloskop zeigt dagegen den vollständigen zeitlichen Verlauf und macht Phasenmessungen möglich.

---

## Wechselstromwiderstände

Es gibt drei Arten von Wechselstromwiderständen:  
- Ohmscher Widerstand (keine Phasenverschiebung)  
- Kondensator (Strom eilt um $+90^\circ$ voraus)  
- Spule (Strom hinkt um $-90^\circ$ hinterher)

Die komplexen Impedanzen lauten:

$$
Z_R = R,
\qquad
Z_C = -\frac{i}{\omega C},
\qquad
Z_L = i\omega L.
$$

Allgemein gilt für die Impedanz:

$$
Z = \frac{U_0}{I_0} e^{i\Delta\varphi}.
$$

Die Phasenverschiebung ergibt sich aus:

$$
\tan(\Delta\varphi) = \frac{\operatorname{Im}(Z)}{\operatorname{Re}(Z)}.
$$

---

## Serienschaltung von R₀ und unbekannter Impedanz X

Die untersuchte Schaltung enthält einen bekannten Widerstand $R_0$ und eine unbekannte Impedanz $X$ (Spule oder Kondensator) in Reihe:

$$
Z_\text{ges} = R_0 + iX.
$$

Der Betrag der Gesamtimpedanz:

$$
|Z_\text{ges}| = \sqrt{R_0^2 + X^2}.
$$

Der Strom ist überall gleich groß. Daher gilt:

$$
I_0 = \frac{U_1}{|Z_\text{ges}|} = \frac{U_2}{R_0},
$$

wobei $U_1$ die Gesamtspannung und $U_2$ die Spannung an $R_0$ ist.

Daraus folgt für den Betrag der unbekannten Impedanz:

$$
|X| = R_0 \sqrt{\left(\frac{U_1}{U_2}\right)^2 - 1}.
$$

Die Phasenverschiebung ergibt sich durch:

$$
\Delta\varphi = \arctan\left(\frac{X}{R_0}\right).
$$

Für Spulen ist $X = +X_L$, für Kondensatoren $X = -X_C$.

---

## Spezialfall: rein ohmischer zweiter Widerstand

Wenn die unbekannte Impedanz ebenfalls ohmisch ist ($X = 0$), gilt:

$$
R_x = R_0\left(\frac{U_1}{U_2} - 1\right),
\qquad
\Delta\varphi = 0.
$$

---

## Phasenmessung am Oszilloskop

Die Spannung $U_2$ ist immer in Phase mit dem Strom, da an einem ohmschen Widerstand keine Phasenverschiebung entsteht.  
Die Zeitdifferenz $\Delta t$ zwischen den Nulldurchgängen von $U_1(t)$ und $U_2(t)$ entspricht daher der Phasenverschiebung zwischen Strom und Gesamtspannung.

Umrechnung:

$$
\Delta\varphi = 2\pi \frac{\Delta t}{T}.
$$

Vorzeichenregel: Eilt $U_2$ voraus, ist die Phasenverschiebung negativ.

Im XY-Betrieb erscheinen typische Lissajous-Figuren:   
- Gerade → rein ohmisch  
- Ellipse → kapazitiv oder induktiv  
- geneigte Ellipse → allgemeiner Fall

---

## Bestimmung von Bauteilwerten

Für den Kondensator:

$$
X_C = \frac{1}{\omega C} 
\quad \Longrightarrow \quad 
C = \frac{1}{\omega X_C}.
$$

Für die Spule:

$$
X_L = \omega L
\quad \Longrightarrow \quad 
L = \frac{X_L}{\omega}.
$$


```python
import math

# ============================================================
# Rundung auf 5 signifikante Stellen (nur fuer Eingangsangaben)
# ============================================================

def runde_5(x):
    """
    Rundet eine Zahl x auf 5 signifikante Stellen.
    Wird nur fuer Groessen OHNE Fehlerangabe genutzt
    (z.B. eingegebene Messwerte).
    """
    if x == 0:
        return 0.0
    stelle = 5 - int(math.floor(math.log10(abs(x)))) - 1
    return round(x, stelle)

# ============================================================
# 5 prozent Unsicherheit und gemeinsame Rundung von Wert und Fehler
# ============================================================

def unsicherheit_5_prozent(wert):
    return abs(wert) * 0.05

def runde_wert_und_fehler(wert, fehler):
    """
    Rundet Wert und Fehler nach ueblichen Labor-Konventionen:

    - Fehler bekommt 1 (oder 2) signifikante Stellen:
        * wenn erste Ziffer 1 oder 2 -> 2 signifikante Stellen
        * sonst -> 1 signifikante Stelle
    - Wert wird auf die gleiche Dezimalstelle gerundet wie der Fehler

    Rueckgabe:
        (wert_gerundet, fehler_gerundet)
    """
    if fehler == 0 or not math.isfinite(fehler):
        # kein sinnvoller Fehler -> Wert normal auf 5 signifikante Stellen
        return runde_5(wert), 0.0

    fehler_abs = abs(fehler)

    # Exponent des Fehlers bestimmen
    exponent = int(math.floor(math.log10(fehler_abs)))
    # erste Ziffer
    erste_ziffer = fehler_abs / (10.0 ** exponent)

    # Anzahl signifikanter Stellen fuer den Fehler
    if erste_ziffer < 3.0:
        sig = 2  # 2 signifikante Stellen (z.B. 1.2, 2.3)
    else:
        sig = 1  # 1 signifikante Stelle (z.B. 4, 5, 8)

    # Anzahl Dezimalstellen fuer round()
    dezimalstellen = sig - 1 - exponent
    # Fehler runden
    fehler_gerundet = round(fehler, dezimalstellen)
    # Wert auf gleiche Dezimalstelle runden
    wert_gerundet = round(wert, dezimalstellen)

    return wert_gerundet, fehler_gerundet

# ============================================================
# Physikalische Grundfunktionen
# ============================================================

def berechne_ohmschen_zusatzwiderstand(R_bekannt, U_gesamt, U_teil):
    return R_bekannt * (U_gesamt / U_teil - 1)


def betragsreaktanz(R_bekannt, U_gesamt, U_teil):
    r = U_gesamt / U_teil
    q = r * r
    if q < 1.0:
        return 0.0
    return R_bekannt * math.sqrt(q - 1)


def phase_theorie(R_bekannt, X):
    phi = math.atan2(X, R_bekannt)
    phi_deg = phi * 180.0 / math.pi
    return phi, phi_deg


def phase_aus_zeitdifferenz(dt, freq, vorzeichen):
    """
    delta_phi = 2*pi*freq*dt
    Vorzeichen: +1 (induktiv), -1 (kapazitiv), 0 (Betrag)
    """
    phi_abs = 2 * math.pi * freq * dt

    if vorzeichen > 0:
        phi = +phi_abs
    elif vorzeichen < 0:
        phi = -phi_abs
    else:
        phi = phi_abs

    phi_deg = phi * 180.0 / math.pi
    return phi, phi_deg


def kapazitaet_aus_Xc(freq, Xc):
    omega = 2 * math.pi * freq
    return 1.0 / (omega * Xc)


def induktivitaet_aus_Xl(freq, Xl):
    omega = 2 * math.pi * freq
    return Xl / omega

# ============================================================
# Auswertung
# mode = "R", "C", "L"
# ============================================================

def auswertung(mode, R_bekannt, U_gesamt, U_teil, freq, dt=None):
    print("============================================================")
    print("Modus =", mode)
    print("R_bekannt =", runde_5(R_bekannt), "Ohm")
    print("U_gesamt =", runde_5(U_gesamt), "V")
    print("U_teil =", runde_5(U_teil), "V")
    print("Frequenz =", runde_5(freq), "Hz")
    print()

    # ----------------------------------------------------------
    # Ohmscher Fall
    # ----------------------------------------------------------
    if mode == "R":
        Rx = berechne_ohmschen_zusatzwiderstand(R_bekannt, U_gesamt, U_teil)
        Rx_err = unsicherheit_5_prozent(Rx)
        Rx_r, Rx_err_r = runde_wert_und_fehler(Rx, Rx_err)

        print("Ohmscher Zusatzwiderstand:")
        print("Rx =", Rx_r, "Ohm +/-", Rx_err_r, "Ohm")
        print("XY: Gerade")

        if dt is not None:
            phi, phi_deg = phase_aus_zeitdifferenz(dt, freq, vorzeichen=0)
            phi_err = unsicherheit_5_prozent(phi)
            phi_deg_err = unsicherheit_5_prozent(phi_deg)
            phi_r, phi_err_r = runde_wert_und_fehler(phi, phi_err)
            phi_deg_r, phi_deg_err_r = runde_wert_und_fehler(phi_deg, phi_deg_err)

            print()
            print("Phasenverschiebung:")
            print("phi =", phi_r, "rad +/-", phi_err_r, "rad")
            print("phi =", phi_deg_r, "grad +/-", phi_deg_err_r, "grad")
        

    # ----------------------------------------------------------
    # Kapazitiver Fall
    # ----------------------------------------------------------
    elif mode == "C":
        Xabs = betragsreaktanz(R_bekannt, U_gesamt, U_teil)
        Xabs_err = unsicherheit_5_prozent(Xabs)
        X = -Xabs

        Xabs_r, Xabs_err_r = runde_wert_und_fehler(Xabs, Xabs_err)

        print("Reaktanz (kapazitiv):")
        print("|X| =", Xabs_r, "Ohm +/-", Xabs_err_r, "Ohm")

        C = None
        if Xabs > 0:
            C = kapazitaet_aus_Xc(freq, Xabs)
            C_err = unsicherheit_5_prozent(C)
            C_r, C_err_r = runde_wert_und_fehler(C, C_err)
            print("C =", C_r, "F +/-", C_err_r, "F")
        else:
            print("C kann nicht berechnet werden (|X| ~ 0).")

        phi_th, phi_th_deg = phase_theorie(R_bekannt, X)
        phi_th_err = unsicherheit_5_prozent(phi_th)
        phi_th_deg_err = unsicherheit_5_prozent(phi_th_deg)
        phi_th_r, phi_th_err_r = runde_wert_und_fehler(phi_th, phi_th_err)
        phi_th_deg_r, phi_th_deg_err_r = runde_wert_und_fehler(phi_th_deg, phi_th_deg_err)

        print()
        print("Theoretische Phasenverschiebung:")
        print("phi_theorie =", phi_th_r, "rad +/-", phi_th_err_r, "rad")
        print("phi_theorie =", phi_th_deg_r, "grad +/-", phi_th_deg_err_r, "grad")

        if dt is not None:
            phi_m, phi_m_deg = phase_aus_zeitdifferenz(dt, freq, vorzeichen=-1)
            phi_m_err = unsicherheit_5_prozent(phi_m)
            phi_m_deg_err = unsicherheit_5_prozent(phi_m_deg)
            phi_m_r, phi_m_err_r = runde_wert_und_fehler(phi_m, phi_m_err)
            phi_m_deg_r, phi_m_deg_err_r = runde_wert_und_fehler(phi_m_deg, phi_m_deg_err)

            diff_deg = phi_m_deg - phi_th_deg
            # Abweichung als "Wert ohne Fehler", aber gerundet
            diff_deg_r = runde_5(diff_deg)

            print()
            print("Gemessene Phasenverschiebung:")
            print("phi_mess =", phi_m_r, "rad +/-", phi_m_err_r, "rad")
            print("phi_mess =", phi_m_deg_r, "grad +/-", phi_m_deg_err_r, "grad")
            print("Abweichung mess - theorie =", diff_deg_r, "grad")
        print("XY: Ellipse")

    # ----------------------------------------------------------
    # Induktiver Fall
    # ----------------------------------------------------------
    elif mode == "L":
        Xabs = betragsreaktanz(R_bekannt, U_gesamt, U_teil)
        Xabs_err = unsicherheit_5_prozent(Xabs)
        X = Xabs

        Xabs_r, Xabs_err_r = runde_wert_und_fehler(Xabs, Xabs_err)

        print("Reaktanz (induktiv):")
        print("|X| =", Xabs_r, "Ohm +/-", Xabs_err_r, "Ohm")

        L = None
        if Xabs > 0:
            L = induktivitaet_aus_Xl(freq, Xabs)
            L_err = unsicherheit_5_prozent(L)
            L_r, L_err_r = runde_wert_und_fehler(L, L_err)
            print("L =", L_r, "H +/-", L_err_r, "H")
        else:
            print("L kann nicht berechnet werden (|X| ~ 0).")

        phi_th, phi_th_deg = phase_theorie(R_bekannt, X)
        phi_th_err = unsicherheit_5_prozent(phi_th)
        phi_th_deg_err = unsicherheit_5_prozent(phi_th_deg)
        phi_th_r, phi_th_err_r = runde_wert_und_fehler(phi_th, phi_th_err)
        phi_th_deg_r, phi_th_deg_err_r = runde_wert_und_fehler(phi_th_deg, phi_th_deg_err)

        print()
        print("Theoretische Phasenverschiebung:")
        print("phi_theorie =", phi_th_r, "rad +/-", phi_th_err_r, "rad")
        print("phi_theorie =", phi_th_deg_r, "grad +/-", phi_th_deg_err_r, "grad")

        if dt is not None:
            phi_m, phi_m_deg = phase_aus_zeitdifferenz(dt, freq, vorzeichen=+1)
            phi_m_err = unsicherheit_5_prozent(phi_m)
            phi_m_deg_err = unsicherheit_5_prozent(phi_m_deg)
            phi_m_r, phi_m_err_r = runde_wert_und_fehler(phi_m, phi_m_err)
            phi_m_deg_r, phi_m_deg_err_r = runde_wert_und_fehler(phi_m_deg, phi_m_deg_err)

            diff_deg = phi_m_deg - phi_th_deg
            diff_deg_r = runde_5(diff_deg)

            print()
            print("Gemessene Phasenverschiebung:")
            print("phi_mess =", phi_m_r, "rad +/-", phi_m_err_r, "rad")
            print("phi_mess =", phi_m_deg_r, "grad +/-", phi_m_deg_err_r, "grad")
            print("Abweichung mess - theorie =", diff_deg_r, "grad")
        print("XY: Ellipse")

    print("============================================================")
    print()
```
