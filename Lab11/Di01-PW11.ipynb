# Elektrische Schwingungen im RLC-Schwingkreis – Gesamte Theorie

Ein elektrischer Schwingkreis besteht aus einer Induktivität L, einer Kapazität C und einem
ohmschen Widerstand R. Er ist ein lineares physikalisches System, in dem Energie zwischen
elektrischem und magnetischem Feld ausgetauscht wird. Der Kondensator speichert
elektrische Energie, die Spule magnetische Energie, während der Widerstand zu
Energieverlusten führt.

Die im Kondensator gespeicherte elektrische Energie beträgt
$$
E_{el} = \frac{q^2}{2C}
$$

Die in der Spule gespeicherte magnetische Energie beträgt
$$
E_{mag} = \frac{L I^2}{2}
$$

Für den geschlossenen Schwingkreis gilt die Kirchhoffsche Maschenregel
$$
U_L + U_R - U_C = 0
$$

Mit
$$
U_L = L \frac{dI}{dt}, \quad
U_R = R I, \quad
U_C = \frac{q}{C}, \quad
I = \frac{dq}{dt}
$$

ergibt sich die Bewegungsgleichung
$$
L \frac{d^2 q}{dt^2} + R \frac{dq}{dt} + \frac{1}{C} q = 0
$$

Nach Division durch L erhält man
$$
\frac{d^2 q}{dt^2} + 2\delta \frac{dq}{dt} + \omega_0^2 q = 0
$$

mit
$$
\delta = \frac{R}{2L}, \qquad
\omega_0 = \frac{1}{\sqrt{LC}}
$$

Für den unterdämpften Fall gilt
$$
q(t) = q_0 e^{-\delta t} \cos(\omega_d t + \varphi), \qquad
\omega_d = \sqrt{\omega_0^2 - \delta^2}
$$

Die Amplitude nimmt exponentiell ab:
$$
A(t) = A_0 e^{-\delta t}, \qquad
\tau = \frac{1}{\delta}
$$

Der Gütefaktor ist definiert als
$$
Q = \frac{\omega_0}{2\delta}
$$

Wird der Schwingkreis durch eine äußere Wechselspannung
$$
U(t) = U_0 \cos(\omega t)
$$

angeregt, so ergibt sich
$$
\frac{d^2 U_C}{dt^2}
+ 2\delta \frac{dU_C}{dt}
+ \omega_0^2 U_C
= \frac{U_0}{LC} \cos(\omega t)
$$

Die stationäre Lösung lautet
$$
U_C(t) = U_{C0}(\omega) \cos(\omega t + \varphi)
$$

mit der Resonanzkurve
$$
U_{C0}(\omega)
=
\frac{U_0 \omega_0^2}
{\sqrt{(\omega_0^2 - \omega^2)^2 + 4\delta^2 \omega^2}}
$$

Die Phasenverschiebung ist gegeben durch
$$
\tan(\varphi) =
\frac{2\delta \omega}{\omega^2 - \omega_0^2}
$$


# Experiment 1 – Freie elektrische Schwingungen

In diesem Experiment wird die freie Schwingung eines RLC-Schwingkreises untersucht.
Der Kondensator wird zunächst auf eine Spannung U₀ geladen. Es gelten die Anfangsbedingungen

$$
q(0) = q_0 = C U_0, \qquad I(0) = 0
$$

Nach dem Schließen des Schwingkreises beginnt der Kondensator sich über die Spule zu
entladen. Die elektrische Energie wird in magnetische Energie umgewandelt und umgekehrt.
Durch den ohmschen Widerstand geht dabei Energie verloren, sodass die Schwingung
gedämpft ist.

Die Bewegungsgleichung lautet
$$
L \frac{d^2 q}{dt^2}
+ R \frac{dq}{dt}
+ \frac{1}{C} q = 0
$$

Mit
$$
\delta = \frac{R}{2L}, \qquad
\omega_0 = \frac{1}{\sqrt{LC}}
$$

ergibt sich für den unterdämpften Fall
$$
q(t) = q_0 e^{-\delta t} \cos(\omega_d t + \varphi)
$$

mit
$$
\omega_d = \sqrt{\omega_0^2 - \delta^2}
$$

Die Amplitude der Schwingung nimmt exponentiell ab:
$$
A(t) = A_0 e^{-\delta t}
$$

Zur Bestimmung der Dämpfung werden die Maxima der Schwingung ausgewertet:
$$
\ln\!\left(\frac{A_i}{A_0}\right) = -\delta t_i
$$

Aus der gemessenen gedämpften Frequenz und der Dämpfung folgt
$$
\omega_0 = \sqrt{\omega_d^2 + \delta^2}
$$

Der Gütefaktor des Schwingkreises ist
$$
Q = \frac{\omega_0}{2\delta}
$$


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# =============================================================================
# PW11 – Experiment 1: Freie Schwingung (Ringdown) – Auswertung (nur OHNE Zusatzwiderstand)
# Eingabe: zwei NumPy-Arrays ti (Zeitstempel der Peaks) und Ai (Peak-Amplituden)
# Unsicherheiten: u(t) = 3% von t, u(A) = 3% von A
# Ausgabe strikt nummeriert (1–5) und Plots getrennt, jeweils mit R^2 im Plot
#
# Hinweis zur Notation:
#   ε (epsilon) aus manchen Skripten ist hier identisch zu δ (delta), der Dämpfungskonstante [1/s].
# =============================================================================

# =========================
# KONFIGURATION
# =========================
L = 1e-3               # Induktivität [H] (nur für R_L und C)
REL_U_T = 0.03          # relative Unsicherheit von t
REL_U_A = 0.03          # relative Unsicherheit von A

# =========================
# DEINE DATEN (hier einsetzen)
# =========================
ti = np.array([0.0, 190e-6, 380e-6, 580e-6, 780e-6, 970e-6, 1.17e-3, 1.37e-3, 1.56e-3, 1.76e-3, 1.96e-3, 2.16e-3], dtype=np.double) # z.B. np.array([...])  Zeitpunkte der Maxima [s]
Ai = np.array([7.6e-3, 7.2e-3, 6.4e-3, 5.8e-3, 5e-3, 4.6e-3, 4.2e-3, 3.6e-3, 3.2e-3, 3e-3, 2.8e-3, 2.4e-3], dtype=np.double)  # z.B. np.array([...])  Amplituden der Maxima [V]
print(ti)
print(Ai)
# =========================
# VORBEREITUNG + UNSICHERHEITEN
# =========================
#ti = np.asarray(ti, dtype=float)
#Ai = np.asarray(Ai, dtype=float)

u_t = REL_U_T * np.abs(ti)
u_A = REL_U_A * np.abs(Ai)

t0 = ti[0]
A0 = Ai[0]

# =========================
# (1) PERIODENDAUER Td UND omega_d
# =========================
dT = np.diff(ti)
u_dT = np.sqrt(u_t[1:]**2 + u_t[:-1]**2)

Td = np.mean(dT)
# Unsicherheit des Mittelwerts: Var(mean) = sum Var(dT_i) / n^2 (Unabhängigkeit angenommen)
u_Td = np.sqrt(np.sum(u_dT**2)) / len(dT)

omega_d = 2.0 * np.pi / Td
u_omega_d = (2.0 * np.pi / Td**2) * u_Td

# =========================
# (2) DÄMPFUNGSKONSTANTE δ AUS ln(A/A0) = -δ (t-t0)
#     -> gewichteter linearer Fit (nur y-Unsicherheiten; x-Unsicherheiten werden vernachlässigt)
# =========================
x = ti - t0
y = np.log(Ai / A0) 
print("x:", x)
print("y:", y)

# Unsicherheit von y = ln(Ai) - ln(A0): u_y^2 = (u_Ai/Ai)^2 + (u_A0/A0)^2
u_y = np.sqrt((u_A / Ai)**2 + (u_A[0] / A0)**2)

def lin_mod(x, m, b):
    return m * x + b

#p, cov = np.polyfit(x, y, deg=1, cov=True)
popt, pcov = curve_fit(lin_mod, x, y, sigma=u_y, absolute_sigma=True)
m_fit, b_fit = popt
u_m = np.sqrt(pcov[0, 0])
#u_m = np.sqrt(cov[0,0])

delta = -m_fit
u_delta = u_m

# R^2 (für den Fit in (x,y))
y_hat = lin_mod(x, m_fit, b_fit)
ss_res = np.sum((y - y_hat)**2)
ss_tot = np.sum((y - np.mean(y))**2)
R2_log = 1.0 - ss_res / ss_tot if ss_tot > 0 else np.nan

# =========================
# (3) EIGENFREQUENZ omega0:  omega_d^2 = omega0^2 - δ^2  -> omega0 = sqrt(omega_d^2 + δ^2)
# =========================
omega0 = np.sqrt(omega_d**2 + delta**2)

# Fehlerfortpflanzung:
# d omega0 = (omega_d/omega0) d omega_d  + (delta/omega0) d delta  (quadratisch addiert)
u_omega0 = np.sqrt((omega_d/omega0 * u_omega_d)**2 + (delta/omega0 * u_delta)**2)

# =========================
# (4) Spulenwiderstand R_L und Kapazität C (ohne Zusatzwiderstand)
#     δ = R_L / (2L)  -> R_L = 2 L δ
#     omega0 = 1/sqrt(LC) -> C = 1/(L omega0^2)
# =========================
R_L = 2.0 * L * delta
u_R_L = 2.0 * L * u_delta

C = 1.0 / (L * omega0**2)
# dC/C = 2 * d(omega0)/omega0  (L als konstant angenommen)
u_C = C * 2.0 * (u_omega0 / omega0)

# =========================
# (5) Güte Q = omega0 / (2 δ)
# =========================
Q = omega0 / (2.0 * delta)
# Fehlerfortpflanzung: dQ/Q = sqrt( (dω0/ω0)^2 + (dδ/δ)^2 )
u_Q = Q * np.sqrt((u_omega0/omega0)**2 + (u_delta/delta)**2)

# =========================
# KONSOLEN-AUSGABE (nummeriert 1–5)
# =========================
print("PW11 – Experiment 1: Freie Schwingung (Ringdown) – Auswertung (ohne Zusatzwiderstand)\n")

print("1) Periodendauer und gedämpfte Kreisfrequenz")
print(f"   Td = ({Td:.6e} ± {u_Td:.1e}) s")
print(f"   ω_d = ({omega_d:.6e} ± {u_omega_d:.1e}) rad/s\n")

print("2) Dämpfungskonstante δ (≙ ε)")
print(f"   δ = ({delta:.6e} ± {u_delta:.1e}) 1/s")
print(f"   R² (Linearfit ln(A/A0) vs. t) = {R2_log:.5f}\n")

print("3) Eigenkreisfrequenz ω0")
print(f"   ω0 = ({omega0:.6e} ± {u_omega0:.1e}) rad/s\n")

print("4) Spulenwiderstand R_L und Kapazität C (ohne Zusatzwiderstand)")
print(f"   R_L = ({R_L:.6e} ± {u_R_L:.1e}) Ω")
print(f"   C   = ({C:.6e} ± {u_C:.1e}) F\n")

print("5) Güte Q")
print(f"   Q = ({Q:.6e} ± {u_Q:.1e})\n")

# =========================
# PLOT: Dämpfungs-Fit ln(A/A0) vs (t-t0) mit Fehlerbalken + R² + δ
# =========================
x_fit = np.linspace(0, x.max(), 400)
y_fit = lin_mod(x_fit, m_fit, b_fit)

fig, ax = plt.subplots(figsize=(7, 4))
ax.errorbar(x, y, yerr=u_y, fmt="o", capsize=3, label="Messwerte: ln(A/A0)")
ax.plot(x_fit, y_fit, "-", label="Gewichteter Linearfit")
ax.set_xlabel("t − t0 [s]")
ax.set_ylabel("ln(A / A0)")
ax.set_title("Dämpfungsfit (Linearisiert)")
ax.grid(alpha=0.3)
ax.legend()
ax.text(
    0.03, 0.05,
    f"R² = {R2_log:.5f}\nδ = ({delta:.3e} ± {u_delta:.1e}) 1/s",
    transform=ax.transAxes,
    va="bottom", ha="left",
    bbox=dict(boxstyle="round,pad=0.35", facecolor="white", alpha=0.9)
)
plt.tight_layout()
plt.show()
```

Die kritische Dämpfung ist 55.4 Ohm (tritt keine Schwingung ein).

# Experiment 2 – Erzwungene Schwingungen und Resonanz

In diesem Experiment wird der RLC-Schwingkreis durch eine äußere sinusförmige Spannung
angeregt. Die Anregung erfolgt mit der Kreisfrequenz ω:

$$
U(t) = U_0 \cos(\omega t)
$$

Die Bewegungsgleichung für die Kondensatorspannung lautet
$$
\frac{d^2 U_C}{dt^2}
+ 2\delta \frac{dU_C}{dt}
+ \omega_0^2 U_C
= \frac{U_0}{LC} \cos(\omega t)
$$

Nach dem Einschwingvorgang bleibt eine stationäre Schwingung der Form
$$
U_C(t) = U_{C0}(\omega) \cos(\omega t + \varphi)
$$

Die frequenzabhängige Amplitude ergibt die Resonanzkurve
$$
U_{C0}(\omega)
=
\frac{U_0 \omega_0^2}
{\sqrt{(\omega_0^2 - \omega^2)^2 + 4\delta^2 \omega^2}}
$$

Das Maximum der Resonanz tritt bei
$$
\omega_{max} = \sqrt{\omega_0^2 - 2\delta^2}
$$

auf und ist für schwache Dämpfung näherungsweise gleich der Eigenkreisfrequenz.

Die Halbwertsbreite der Resonanzkurve beträgt
$$
\Delta \omega = 2\delta
$$

Daraus folgt für den Gütefaktor
$$
Q \approx \frac{\omega_{max}}{\Delta \omega}
$$

Die Phasenverschiebung zwischen Anregung und Antwort ist
$$
\tan(\varphi) =
\frac{2\delta \omega}{\omega^2 - \omega_0^2}
$$

Beim Durchgang durch die Resonanz ändert sich die Phase charakteristisch.


```python
"""
PW11 - Experiment 2: Erzwungene Schwingungen (Parallelresonanz) Auswertung
Ausgabe nummeriert nach Aufgabenstellung 2.2

Eingabe (Messung):
1) Resonanzkurve: f_curve (Hz), UC0_curve (V)  [punktweise gemessen]
2) Phase: f_phi (Hz) und delta_t (s) zwischen UC und U (Cursor-Messung)

Auswertung:
- Umrechnung f -> omega = 2*pi*f
- Bestimmung von omega_max und U_max
- Bestimmung der 1/sqrt(2)-Breite: Delta_omega = omega_2 - omega_1 bei U_max/sqrt(2)
- Daempfung: delta = Delta_omega/2
- Guete: Q approx omega_max / Delta_omega
- Eigenfrequenz (Parallelresonanz): omega_max = sqrt(omega_0^2 + 2*delta^2) -> omega_0 = sqrt(omega_max^2 - 2*delta^2)

Plot:
- Nur Resonanzkurve (Messpunkte + Linie)
- Textbox unten links: nur R^2 (Qualitaet eines einfachen Fits an ein glattes Modell)

Benötigt: numpy, matplotlib, scipy
"""

import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit


# =========================================================
# HIER DEINE MESSDATEN EINTRAGEN
# =========================================================

# Resonanzkurve (punktweise gemessen)
f_curve = np.array(
    [2580, 2880, 3180, 3480, 3780, 4080, 4380, 4680, 4980, 5280, 5580, 5880, 6180, 6480, 6780,
     3000, 4590, 4720, 5090, 5000, 5180, 5050, 5130],
    dtype=float
)  # Hz

UC0_curve = np.array(
    [2.6e-3, 3.4e-3, 4.8e-3, 7e-3, 10e-3, 15.1e-3, 24e-3, 47.6e-3, 164e-3, 106e-3, 50e-3,
     34.4e-3, 26.8e-3, 22.8e-3, 19.6e-3, 7.6e-3, 37.6e-3, 52.8e-3, 262e-3, 188e-3,
     174e-3, 246e-3, 226e-3],
    dtype=float
)  # V

# Phase-Messung: f_phi (Hz) und delta_t (s)
# In der Praxis: f_phi sind genau die Frequenzen, bei denen du delta_t am Oszi abliest.
f_phi = np.array([4072.0, 5090.0, 6110.0], dtype=float)    # Hz
delta_t = np.array([131e-6, 44e-6, 4e-6], dtype=float)      # s


# =========================================================
# MODEL + HILFSFUNKTIONEN
# =========================================================

def model_parallel_resonance(omega, K, omega0, delta):
    # Formel wie in deinem Code (NICHT veraendert)
    denom = np.sqrt((omega0**2 - omega**2)**2 + (2.0 * delta * omega)**2)
    return K * (omega**2) / denom


def r2_score(y, y_hat):
    ss_res = float(np.sum((y - y_hat) ** 2))
    ss_tot = float(np.sum((y - np.mean(y)) ** 2))
    return float(1.0 - ss_res / ss_tot) if ss_tot > 0 else np.nan


def estimate_half_power_bandwidth(omega, U):
    # Bestimme omega_max und die beiden Schnittpunkte bei U_max/sqrt(2)
    idx_max = int(np.argmax(U))
    omega_max = float(omega[idx_max])
    U_max = float(U[idx_max])
    U_half = U_max / np.sqrt(2.0)

    left_omega = np.nan
    right_omega = np.nan

    # Links suchen
    for i in range(idx_max - 1, -1, -1):
        if (U[i] - U_half) <= 0.0 and (U[i + 1] - U_half) >= 0.0:
            x0, y0 = omega[i], U[i]
            x1, y1 = omega[i + 1], U[i + 1]
            left_omega = float(x0 + (U_half - y0) * (x1 - x0) / (y1 - y0))
            break

    # Rechts suchen
    for i in range(idx_max, len(U) - 1):
        if (U[i] - U_half) >= 0.0 and (U[i + 1] - U_half) <= 0.0:
            x0, y0 = omega[i], U[i]
            x1, y1 = omega[i + 1], U[i + 1]
            right_omega = float(x0 + (U_half - y0) * (x1 - x0) / (y1 - y0))
            break

    Delta_omega = float(right_omega - left_omega) if np.isfinite(left_omega) and np.isfinite(right_omega) else np.nan
    return omega_max, U_max, U_half, left_omega, right_omega, Delta_omega


def compute_phi_from_dt(f_hz, dt_s):
    omega = 2.0 * np.pi * f_hz
    phi = omega * dt_s  # rad
    # auf [0, pi] bringen (praktisch fuer Darstellung)
    phi_mod = np.mod(phi, 2.0 * np.pi)
    phi_mod = np.where(phi_mod > np.pi, 2.0 * np.pi - phi_mod, phi_mod)
    return omega, phi, phi_mod


def pretty_rad(x):
    return f"{x:.6e} rad/s"


# =========================================================
# AUSWERTUNG
# =========================================================

# Sortieren (sehr wichtig! sonst wird die Halbwertsuche/Interpolation falsch)
order = np.argsort(f_curve)
f_curve = f_curve[order]
UC0_curve = UC0_curve[order]
omega_curve = 2.0 * np.pi * f_curve

# 1) Empirisch: omega_max und Bandbreite
omega_max, U_max, U_half, omega_1, omega_2, Delta_omega = estimate_half_power_bandwidth(omega_curve, UC0_curve)

# 3) delta, Q, omega0 (Parallelresonanz-Formel)
delta_emp = float(Delta_omega / 2.0) if np.isfinite(Delta_omega) else np.nan
Q_emp = float(omega_max / Delta_omega) if np.isfinite(Delta_omega) and Delta_omega > 0 else np.nan

omega0_emp = np.nan
if np.isfinite(delta_emp) and np.isfinite(omega_max):
    val = omega_max**2 - 2.0 * (delta_emp**2)
    omega0_emp = float(np.sqrt(val)) if val > 0 else np.nan

# 2) Phase
omega_phi, phi_raw, phi_mod = compute_phi_from_dt(f_phi, delta_t)

# =========================================================
# FIT (fix: bessere Startwerte + Bounds + glatte Kurve)
# =========================================================

# bessere Startwerte:
omega0_guess = omega_max if np.isfinite(omega_max) else float(np.median(omega_curve))
delta_guess  = delta_emp if (np.isfinite(delta_emp) and delta_emp > 0) else 300.0

# Bei omega ~ omega0: U_max ≈ K * omega0 / (2*delta)  ->  K ≈ U_max*(2*delta)/omega0
K_guess = float(U_max * (2.0 * delta_guess) / (omega0_guess + 1e-12)) if np.isfinite(U_max) else 1.0

p0 = [K_guess, omega0_guess, delta_guess]
bounds = ([0.0, 0.0, 0.0], [np.inf, np.inf, np.inf])

try:
    popt, _ = curve_fit(
        model_parallel_resonance,
        omega_curve, UC0_curve,
        p0=p0,
        bounds=bounds,
        maxfev=200000
    )
    R2_fit = r2_score(UC0_curve, model_parallel_resonance(omega_curve, *popt))
except Exception as e:
    popt = None
    R2_fit = np.nan
    print("Fit fehlgeschlagen:", e)

# =========================================================
# AUSGABE NACH AUFGABEN 1-3 (PW11, Abschnitt 2.2)
# =========================================================

print("\n===== Aufgabe 1: Resonanzkurve UC0(omega) =====")
print(f"Anzahl Messpunkte: {len(f_curve)}")
print(f"Maximum: omega_max = {pretty_rad(omega_max)},  U_max = {U_max:.6e} V")
print(f"Halbwert: U_max/sqrt(2) = {U_half:.6e} V")
print(f"Schnittpunkte: omega_1 = {pretty_rad(omega_1)} , omega_2 = {pretty_rad(omega_2)}")
print(f"Bandbreite: Delta_omega = {pretty_rad(Delta_omega)}")

print("\n===== Aufgabe 2: Phasenverschiebung phi(omega) =====")
for i in range(len(f_phi)):
    print(
        f"{i+1}) f = {f_phi[i]:.1f} Hz,  delta_t = {delta_t[i]:.6e} s,  "
        f"phi = {phi_mod[i]:.6f} rad"
    )

print("\n===== Aufgabe 3: Eigenfrequenz, Daempfung, Guete aus Resonanzkurve =====")
print(f"delta = Delta_omega/2 = {delta_emp:.6e} 1/s")
print(f"Q approx omega_max/Delta_omega = {Q_emp:.4f}")
print(f"omega_0 aus omega_max = sqrt(omega_0^2 + 2 delta^2) -> omega_0 = {pretty_rad(omega0_emp)}")
print("Hinweis: Vergleich mit Experiment 1 erfolgt durch Gegenueberstellung der dort bestimmten Werte.")

if popt is not None:
    K_fit, omega0_fit, delta_fit = popt
    print("\n===== Fit-Parameter (Modell) =====")
    print(f"K     = {K_fit:.6e}")
    print(f"omega0= {omega0_fit:.6e} rad/s")
    print(f"delta = {delta_fit:.6e} 1/s")

# =========================================================
# PLOT (Resonanzkurve + Textbox unten links: nur R^2)
# =========================================================

fig, ax = plt.subplots(figsize=(7.4, 4.6))

ax.scatter(omega_curve, UC0_curve,label="Messpunkte", s=20, color="hotpink")
#ax.plot(omega_curve, UC0_curve, "-", label="Verbindung (nur Darstellung)", linewidth=1.2, alpha=0.6)

# Fitkurve auf dichtem Grid -> sieht dann "richtig" aus
if popt is not None:
    omega_dense = np.linspace(float(np.min(omega_curve)), float(np.max(omega_curve)), 2500)
    ax.plot(omega_dense, model_parallel_resonance(omega_dense, *popt), "--", label="Modell-Fit", linewidth=2)

# Markierungen: omega_max und Halbwertpunkte (falls vorhanden)
ax.axvline(omega_max, linestyle=":", linewidth=1.5, label="omega_max")
if np.isfinite(omega_1):
    ax.axvline(omega_1, linestyle=":", linewidth=1.0)
if np.isfinite(omega_2):
    ax.axvline(omega_2, linestyle=":", linewidth=1.0)

ax.set_xlabel("omega (rad/s)")
ax.set_ylabel("UC0 (V)")
ax.set_title("Resonanzkurve: Parallelresonanz (Experiment 2)")
ax.grid(alpha=0.25)
ax.legend(loc="upper right", frameon=True)

box_text = f"Fit-Qualitaet\nR^2 = {R2_fit:.5f}" if np.isfinite(R2_fit) else "Fit-Qualitaet\nR^2 = n/a"
ax.text(
    0.03, 0.06, box_text,
    transform=ax.transAxes,
    va="bottom",
    ha="left",
    fontsize=10,
    bbox=dict(boxstyle="round,pad=0.35", facecolor="white", edgecolor="0.75", alpha=0.92)
)

plt.tight_layout()
plt.show()
```
